## Task 3: Data Cleansing and Transformation

### **Missing values, invalid values, inconsistent categories, incorrect data types**  
They have been identified in Task 2, none need to be addressed  


```python
# Confirm duplicate count again
print(f"Duplicate rows before cleaning: {df.duplicated().sum()}")

#Check the class distribution within the duplicates specifically
duplicates = df[df.duplicated(keep=False)]
print(duplicates['Class'].value_counts())

#Remove duplicates
df_cleaned = df.drop_duplicates().copy()
print(f"Shape before cleaning: {df.shape}")
print(f"Shape after cleaning: {df_cleaned.shape}")
print(f"Duplicate rows after cleaning: {df_cleaned.duplicated().sum()}")
```

### Duplicate Records  
1,081 duplicate rows were identified in the dataset (1,822 rows total were involved in duplication before removal). From these, 32 belonged to the fraud class (Class = 1), which were approximately 6.5% of all fraud cases in the dataset.  
As said in Task 2, there is no transaction ID or any other identifiers known to confirm whether or not these duplicate rows represent real repeated transactions or atefacts of data collection. Given this, duplicates were removed using drop_duplicates(), reducing the dataset from 284,807 to 283,726 rows.  
This decision was made mainly to avoid data leaks: if identical rows were split across training and test sets, the model would be evaluated on its already seen, artificially infalting performance metrics. This is to outweigh the risk of discarding genuine fraud attempts, since PCA-transformed features being identical across two rows makes coincidental duplication in this many dimensions extremely unlikely for two actual distinct transactions.

```python
import matplotlib.pyplot as plt

#Boxplot to visualise outliers in Amount
plt.figure(figsize=(8,4))
plt.boxplot(df_cleaned['Amount'], vert=False)
plt.title('Boxplot of Transaction Amounts')
plt.xlabel('Amount ($)')
plt.show()

# IQR method to detect outliers
Q1 = df_cleaned['Amount'].quantile(0.25)
Q3 = df_cleaned['Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")

outliers = df_cleaned[(df_cleaned['Amount'] < lower_bound) | (df_cleaned['Amount'] > upper_bound)]
print(f"Number of outliers (IQR method): {outliers.shape[0]}")
print(f"Percentage of dataset: {outliers.shape[0] / df_cleaned.shape[0] * 100:.2f}%")

# Check how many of those outliers are fraud vs legitimate
print(outliers['Class'].value_counts())

# Also check the max value again for context
print(f"Maximum amount: {df_cleaned['Amount'].max()}")
```

```python
# Log-transform Amount to reduce skewness (log1p handles the Amount=0 case safely)
df_cleaned['Amount_log'] = np.log1p(df_cleaned['Amount'])

# Compare skew before and after
print(f"Original Amount skew: {df_cleaned['Amount'].skew():.3f}")
print(f"Log-transformed Amount skew: {df_cleaned['Amount_log'].skew():.3f}")
```

### Outlier Investigation and Treatment  
The Amount feature was investigated for outliers using the IQR method. This flagged 31,685 transactions (11.17% of the dataset) as outliers, with an upper bound of $185.38. Given that over 1 in 10 transactions were flagged, this is interpreted as a sign that the IQR method does not properly suit this feature's right-skewed distribution, rather than evidence of 31,685 invalid data points.  
To confirm these outliers were not primarily fraud-related noise, the class distribution within the outliers was checked: 87 of the 31,685 flagged transactions were fraudulent (0.275%), compared to the overall dataset fraud rate of 0.17%. Fraud is somewhat more common amongst high-value transactions, but the majority (31,598) of flagged outliers are legitimate, high-value purchases.  
Given this, outliers were retained rather than removed or capped. Removing them would have disproportionately discarded fraud cases, going against the project's objective, and there was no evidence that these represented invalid data rather than real large transactions.

### Transformation: Log-Scaling Amount  
Instead of removing outliers, a log transformation(log1p, which appropriately handles Amount = 0 cases) was used to counteract the right-skew of the Amount feature. This reduced the skewness from 16.979 to 0.161, bringing the distribution close to symmetry. This transformation preserves all outlier information whilst making the feature far more suitable for machine learning models that assume or perform better with roughly normally-distributed inputs.


```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

#Scale Time and Amount_log
df_cleaned['Time_scaled'] = scaler.fit_transform(df_cleaned[['Time']])
df_cleaned['Amount_scaled'] = scaler.fit_transform(df_cleaned[['Amount_log']])

#Quick check: compare before/after ranges
print(df_cleaned[['Time', 'Time_scaled', 'Amount_log', 'Amount_scaled']].describe())
```

### Scaling Time and Amount  
Unlike V1-V28, which are already approximately standardised as a result of the PCA transformation (mean = 0), the Time and Amount_log features were on differing numeric scales (Time ranging up to aprox. 172,792; Amount_log ranging 0-10). Since many machine learning algorithims are sensitive to differences in feature scale, StandardScaler was applied to both features independently, transforming each to have a mean of 0 and a standard deviation of 1.  
After scaling, Time_scaled ranges from approximately -2.00 to 1.64, and Amount_scaled ranges from approximately -1.90 to 4.22, bringing both features onto a comparable numeric footing with the existing V1-V28 components.

```python
# Define final feature set for modelling: drop raw/columns, keep scaled versions
final_columns = ['Time_scaled'] + [f'V{i}' for i in range(1, 29)] + ['Amount_scaled', 'Class']
df_final = df_cleaned[final_columns].copy()

print(df_final.shape)
df_final.head()
```

### Final Cleaned Dataset  
After cleaning and transformation, the final dataset (df_final) contains 283,726 rows and 31 columns: Time_scaled, V1-V28, Amount_scaled, and the target variable Class. This represents the original 284,807 rows with 1,081 duplicates removed, and with Time and Amount replaced by their scaled/transformed equivalents (Time_scaled, Amount_scaled) to ensure all features are on a comparable numeric scaled ahead of modelling.

### Reusable Cleaning Pipeline  

The individual cleaning and transformation steps above (duplicate removal, log transformation, and scaling) are conjoined below into a single reusable function, clean_data(). This allows the entire cleaning process to be reproduced from the untouched raw dataset (df_raw) in one call and makes it easy to compare the raw and cleaned datasets

```python
def clean_data(df_input):
    """
    Cleans and transforms the raw credit card fraud dataset.  
    Steps: remove duplicates, log-transform Amount, scale Time and Amount_log.  
    Returns a new cleaned DataFrame; does not modify the input. 
    """ 
    df=df_input.drop_duplicates().copy()

    df['Amount_log'] = np.log1p(df['Amount'])

    scaler = StandardScaler()
    df['Time_scaled'] = scaler.fit_transform(df[['Time']])
    df['Amount_scaled'] = scaler.fit_transform(df[['Amount_log']])

    final_columns = ['Time_scaled'] + [f'V{i}' for i in range(1, 29)] + ['Amount_scaled', 'Class']
    df_final_check = df[final_columns].copy()
    return df_final_check

# Run the pipeline on the raw copy and confirm it matches the manual cleaning
df_final_check = clean_data(df_raw)

print(f"Raw shape: {df_raw.shape}")
print(f"Cleaned shape (via function): {df_final_check.shape}")
print(f"Matches manual df_final: {df_final_check.equals(df_final)}")
```

## Inferential Statistics  

### Independent Samples T-Test: Transaction Amount by Class

```python
from scipy import stats

#Split Amount_scaled by class
legit_amounts = df_final[df_final['Class'] == 0]['Amount_scaled']
fraud_amounts = df_final[df_final['Class'] == 1]['Amount_scaled']

# Independent samples t-test
t_stat, p_value = stats.ttest_ind(legit_amounts, fraud_amounts, equal_var=False)

print(f"Legitimate transactions: n={len(legit_amounts)}, mean={legit_amounts.mean():.4f}, std={legit_amounts.std():.4f}")
print(f"Fraudulent transactions: n={len(fraud_amounts)}, mean={fraud_amounts.mean():.4f}, std={fraud_amounts.std():.4f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.10f}")
```

### Independent-Samples T-Test: Transaction Amount by Class

-Null hypothesis: There is no difference in mean transaction amount between legitimate and fraudulent transactions.  
-Alternative hypothesis: There is a difference in mean transaction amount between legitimate and fraudulent transactions.  
-Significance level: α = 0.05  

Welch's independent samples t-test was tested comparing Amount_scaled between legitimate (n = 283,253, mean = 0.0003, std = 0.9993) and fraudulent (n = 473, mean = -0.1908, std = 1.3353) transactions.  

What the output shows: t = 3.11, p = 0.00197.  

Main pattern: fraudulent transactions have a lower average amount than legitimate transactions, and show greater variability, consistent with the wider interquartile range seen in the Task 4 boxplot.

Conclusion: since p (0.00197) is below the significance threshold of 0.05, the null hypothesis is rejected. There is statistically significant evidence that the mean transaction amount differs between legitimate and fraudulent transactions.

Limitation: although the result is statistically significant, the difference in means is small (0.0003 vs -0.1908 on a standardised scale). With a very large sample size (283,726 total transactions), t-tests become highly sensitive to even small differences between groups, meaning statistical significance here does not necessarily mean the difference is large or meaningful on its own. This distinction between statistical significance and practical/effect size should be considered before treating Amount as a standalone predictor of fraud. A point also reflected in the relatively weak correlation (-0.008) found between Amount_scaled and Class in the Task 4 correlation analysis.

```python
# Bin Amount into categories to enable a chi-square test
df_final['Amount_bin'] = pd.qcut(df_final['Amount_scaled'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])

# Create a contingency table: Amount_bin vs Class
contingency_table = pd.crosstab(df_final['Amount_bin'], df_final['Class'])
print(contingency_table)

# Chi-square test of independence
chi2, p_value_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\nChi-square statistic: {chi2:.4f}")
print(f"P-value: {p_value_chi2:.10f}")
print(f"Degrees of freedom: {dof}")
```

### Chi-Square Test of Independence: Transaction Amount Category vs Class  

Since the dataset contains no categorical variables, Amount_scaled was binned into four quartile-based categories (Low, Medium, High, Very High) to allow a chi-square test of independence.  

Null hypothesis: Fraud rate is independent of transaction amount category  
Alternative hypothesis: Fraud rate depends on transaction amount category  
Significance level: α = 0.05  

Contingency table:  

Amount category | Legitimate | Fraud  
Low             |  70,727    | 213  
Medium          |  70,988    | 47  
High            |  70,775    | 45  
Very High       |  70,763    | 168  

What the output shows: x^2 = 185.42, p < 0.0001, df = 3  

Main pattern: fraud counts arent evenly distributed across amount categories. The Low and Very High bins contain more fraud cases (213 and 168) than the Medium and High bins (47 and 45), despite each bin having a near-identical number of total transactions (70,000 - 71,000). This suggests fraud is concentrated at both ends of the amount spectrum rather than in the middle range.  

Conclusion: since p < 0.05, the null hypothesis is rejected. There is a lot of statistical evidence that proves fraud rate is associated with transaction amount category.  

Limitation: A significant chi-square result would indicate association, not causation. It does not mean transaction amount causes fraud, only that the two variables are statistically related in this dataset. Since Amount was artifically binned into categories for this test, the specific bin boundaries (quartiles) were a modelling choice rather than natural grouping in the data, and different binning strategies could produce different category-level patterns even if the overall association remains significant
