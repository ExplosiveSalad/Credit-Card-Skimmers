# Task 3: Data Cleaning and Transformation

### Missing values, invalid values, inconsistent categories, incorrect data types, formatting issues
For task 2, all these criteria are met. For non-missing values: Data doesn't contain any non-valid Amount/Time. There's also nothing with conflicting category identifiers (as it does not contain categorical variables), and all numeric types are properly classified.
```python
# Remove duplicate rows, keeping the first occurrence of each
print(f"Rows before deduplication: {df.shape[0]}")
df_cleaned = df.drop_duplicates(keep='first').copy()
print(f"Rows after deduplication: {df_cleaned.shape[0]}")
print(f"Rows removed: {df.shape[0] - df_cleaned.shape[0]}")
```

### Dupplicate Records
A total of 1,081 duplicate rows existed in the data (prior to cleanup, 1,822 rows had been duplicated). Of those, 32-or approximately 6.5% of all fraud in the dataset-were in the class Fraud (Class = 1).

As mentioned in Task 2, no transaction ID or other identifier existed that could be used to determine whether these duplicate rows were duplicate transactions or data collection artifacts. Consequently, the data set was reduced from 284,807 to 283,726 rows after duplicates were dropped using drop_duplicates().

In order to avoid data leaks, we decided to split identical rows between train and test: if two similar rows fall into these two groups, the classifier would be tested on what it already knew and that would artificially increase performance scores. PCA characteristics are similar for two identical rows, so this kind of coincidence in as many as two points, for two real transactions, is unlikely to happen more than a few times-more dangerous than missing fraud!

```python
import matplotlib.pyplot as plt
from scipy import stats

# Z-score method for outlier detection on Amount
z_scores = np.abs(stats.zscore(df_cleaned['Amount']))
outliers = df_cleaned[z_scores > 3]

print(f"Outliers flagged (|z| > 3): {outliers.shape[0]} ({outliers.shape[0] / len(df_cleaned) * 100:.2f}% of dataset)")
print(outliers['Class'].value_counts())

# Log-scale histogram to visualise the skew that's driving these flags
plt.figure(figsize=(8, 4))
plt.hist(df_cleaned['Amount'], bins=100)
plt.yscale('log')
plt.title('Distribution of Transaction Amount (log-scaled y-axis)')
plt.xlabel('Amount ($)')
plt.ylabel('Count (log scale)')
plt.show()
```

### Outlier Investigation and Treatment

By using a Z-score (transactions where Amount had a value that was >3 std deviations away from the mean, and hence flagged as an outlier), 4,063 outliers were identified, or roughly 1.43% of the entire data set. This is a far less aggressive flagging rule, and much less aggressive than an IQR rule on a distribution so heavily rightward skewed. Z-scores are based on the mean/standard deviation, rather than quartiles, and the heavy rightward skew (skew ~16.98) of Amount greatly magnifies the standard deviation, thus artificially pushing the 3-sigma boundary far outside normal transaction values.

By examining the class split of the outliers that are flagged in particular, 11 of 4,063 are in the fraud class. Only a tiny minority, but enough to suggest that large transaction amounts are not a particularly effective fraud signal on their own. However, because the indication of outlierness in this feature is mostly caused by the natural right skew that everyday spending can often take (people buy occasional large but completely legal items) and not by data errors, no outliers were removed. 

Removing them could mean losing some fraud cases that the model can learn from as well. 

Instead, this skew is handled explicitly through the use of a transformation in the next step. This is generally a much better choice than deletion when outliers are a property of the distribution of a feature rather than a data quality issue.

```python
from sklearn.preprocessing import PowerTransformer, RobustScaler

# Yeo-Johnson power transform on Amount to correct the skew
pt = PowerTransformer(method='yeo-johnson')
df_cleaned['Amount_transformed'] = pt.fit_transform(df_cleaned[['Amount']])

print(f"Original Amount skew: {df_cleaned['Amount'].skew():.3f}")
print(f"Yeo-Johnson transformed Amount skew: {df_cleaned['Amount_transformed'].skew():.3f}")

# RobustScaler on Time, since it scales using the median and IQR rather than
rs = RobustScaler()
df_cleaned['Time_scaled'] = rs.fit_transform(df_cleaned[['Time']])

print(df_cleaned[['Time', 'Time_scaled', 'Amount', 'Amount_transformed']].describe())
```

### Transforming and Scaling Time and Amount
V1-V28 has already been approximately standardized in the PCA transformation; however, the features Time and Amount are on very different scales, and Amount is highly skewed, so both required preprocessing before modeling.

Instead of doing it manually, a Yeo-Johnson power transform was done on Amount. Whereas the log transform requires an offset, the Yeo-Johnson can be applied directly to zero and negative values, and it will work out what the best power parameter to apply is rather than taking 0 as the default log relationship. This gave a reduction in skew of 16.98 to essentially zero (~0.018). 

Time was scaled with a RobustScaler, rather than a z-score scaler which normalizes by subtracting the mean and dividing by the standard deviation. 

RobustScaler normalizes using the median and inter-quartile range, which makes it more robust to the (admittedly few) cases of residual outliers that remained in the data, which was more consistent with a cleaning pipeline that specifically decided not to remove outliers.

```python
# Final feature set for modelling
final_columns = ['Time_scaled'] + [f'V{i}' for i in range(1, 29)] + ['Amount_transformed', 'Class']
df_final = df_cleaned[final_columns].copy()

print(df_final.shape)
df_final.head()
```

### Final Cleaned Dataset
The final dataset (dffinal) comprises 283,726 rows and 31 columns. The columns are Timescaled, V1-V28, Amount_transformed, and the target feature Class. This is the original 284,807 rows with 1,081 duplicates removed and Time and Amount replaced by their scaled/transformed equivalents, so every feature is on a comparable numeric scale before modeling. No rows were eliminated due to outliers. Only the two originally skewed/unscaled features were transformed.

### Reusable Cleaning Pipeline
All above process steps (deduplication, Yeo-Johnson transform, and robust scaling) are summed into a function: The cleaning method used can be applied in a single step using all the provided functions, directly on the clean but raw dataset df_raw.

```python
def clean_data(df_input):
    """
    Cleans and transforms the raw credit card fraud dataset.
    Steps: remove duplicates, Yeo-Johnson transform Amount, robust-scale Time.
    Returns a new cleaned DataFrame; does not modify the input.
    """
    df_out = df_input.drop_duplicates(keep='first').copy()

    pt = PowerTransformer(method='yeo-johnson')
    df_out['Amount_transformed'] = pt.fit_transform(df_out[['Amount']])

    rs = RobustScaler()
    df_out['Time_scaled'] = rs.fit_transform(df_out[['Time']])

    final_cols = ['Time_scaled'] + [f'V{i}' for i in range(1, 29)] + ['Amount_transformed', 'Class']
    return df_out[final_cols].copy()

# Confirm it reproduces df_final from the raw dataset
df_final_check = clean_data(df_raw)
print(df_final_check.shape)
print((df_final_check.values == df_final.values).all())
```

## Infernal Statistics

### Independent-Samples T-Test
- H0 (null hypothesis): The average of the transaction Amount is the same for both fraudulent and legitimate transactions (μ_fraud = μ_legitimate).

- H1 (alternative hypothesis): The mean transaction Amount for fraudulent transactions differs from that for legitimate transactions (fraud ≠ legitimate).

- Alpha level = 0.05

A Welch's t-test ('qual_var=False') is used in place of a regular t-test because the two groups are vastly different in size (473 fraud vs. 283,253 legitimate) and because the variance of Amount varies significantly between the two groups. The Welch's t-test does not assume equal variances.

```python
from scipy import stats

fraud_amount = df_cleaned.loc[df_cleaned['Class'] == 1, 'Amount']
legit_amount = df_cleaned.loc[df_cleaned['Class'] == 0, 'Amount']

t_stat, p_value = stats.ttest_ind(fraud_amount, legit_amount, equal_var=False)

print(f"Fraud:       n={len(fraud_amount)}, mean=${fraud_amount.mean():.2f}, std=${fraud_amount.std():.2f}")
print(f"Legitimate:  n={len(legit_amount)}, mean=${legit_amount.mean():.2f}, std=${legit_amount.std():.2f}")
print(f"\nt-statistic: {t_stat:.4f}")
print(f"p-value: {p_value:.6f}")

alpha = 0.05
if p_value < alpha:
    print(f"\np < {alpha} -> Reject H0: mean transaction Amount differs significantly by Class.")
else:
    print(f"\np >= {alpha} -> Fail to reject H0: no significant difference detected.")
```

### Result: 
Specifically, for our dataset, we have 473 fraudulent with a mean of $123.87,$260.21 std and 283,253 non-fraudulent transactions. These are summarized below. We use Welch's t-test because these samples do not appear to have equal variance (i.e. The sample variances are much larger than the means), consistent with the findings of Task 2, which showed that transactions Amount is highly right-skewed:

t=2.96,  p=0.0032 (<0.05)

Since $p<0.05$, the null hypothesis is rejected, indicating a statistically significant difference in transaction amounts between the two categories of transactions. The fraudulently labelled transactions have, on average, a substantially higher transaction Amount than legitimate ones, however this is potentially misleading - as observed in the Task 4 violin plot - these transactions also highly overlap. This indicates that a statistically significant difference in mean value does not imply a usable predictor in the context of accurately distinguishing between transactions.

### Chi-Square Test of Independence: Transaction Amount Category vs Class

Amount is binned into several bins beforehand since chi-square takes category counts, not averaged numbers.
- H0 (null hypothesis): Amount Category and Class are independent of each other; the category Amount for an online transaction is NOT associated with the fraud status of an online transaction.
- H1 (alternative hypothesis): Amount Category and Class are NOT independent of each other; there IS an association between the fraud status of an online transaction and its category Amount.

- Significance value : α = 0.05

```python
# Bin Amount into categories
bins = [-0.01, 10, 100, 500, df_cleaned['Amount'].max()]
labels = ['Low (<=$10)', 'Medium ($10-$100)', 'High ($100-$500)', 'Very High (>$500)']
df_cleaned['Amount_category'] = pd.cut(df_cleaned['Amount'], bins=bins, labels=labels)

contingency_table = pd.crosstab(df_cleaned['Amount_category'], df_cleaned['Class'])
print("Observed counts:")
print(contingency_table)

print("\nFraud rate within each Amount category:")
print((contingency_table[1] / contingency_table.sum(axis=1) * 100).round(3))

chi2, p_value_chi, dof, expected = stats.chi2_contingency(contingency_table)
print(f"\nchi2 = {chi2:.4f}, dof = {dof}, p-value = {p_value_chi:.6g}")

if p_value_chi < 0.05:
    print("p < 0.05 -> Reject H0: Amount category and Class are significantly associated.")
else:
    print("p >= 0.05 -> Fail to reject H0: no significant association detected.")
```

## Result: 
X^2 = 105.65 (dof = 3), p ~ 9.48*10-23. Well below 0.05 and so reject the null hypothesis. Class is significantly associated with Amount category. 

When you look at the fraud rate within each category, the relationship isn't a simple straight line "Very High" transactions have the highest fraud rate (0.37%), "Low" transactions are second highest (0.24%) whereas "Medium" transactions have the lowest fraud rate (0.09%). 

This tells us that fraud is more concentrated at the extremes of the amount range, rather than simply following a pattern "bigger amount = more likely fraud".
